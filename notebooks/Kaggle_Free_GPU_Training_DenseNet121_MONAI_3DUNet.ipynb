```python
!pip install -q monai[all]==1.3.2 pydicom grad-cam dvc[gdrive]
import torch
import torch.nn as nn
from monai.networks.nets import DenseNet121, UNet
from monai.losses import DiceCELoss
print('PyTorch:', torch.__version__, '| CUDA Available:', torch.cuda.is_available())
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# 1. Initialize Pretrained DenseNet121 for 14-Class Thoracic Pathology Classification
classifier_2d = DenseNet121(spatial_dims=2, in_channels=3, out_channels=14, pretrained=True).to(device)

# 2. Initialize 3D U-Net in MONAI for Volumetric CT/MRI Segmentation
segmenter_3d = UNet(
    spatial_dims=3,
    in_channels=1,
    out_channels=2,
    channels=(16, 32, 64, 128, 256),
    strides=(2, 2, 2, 2),
    num_res_units=2,
).to(device)

torch.save(classifier_2d.state_dict(), 'densenet121_cxr.pth')
torch.save(segmenter_3d.state_dict(), 'unet3d_lung.pth')
print('Exported densenet121_cxr.pth and unet3d_lung.pth — place them in your local models/ folder!')
```
